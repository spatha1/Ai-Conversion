```python
def run_query(query):
    print("Source Connected")

    cursor = source_conn.cursor()
    try:
        cursor.execute(query)
        df = cursor.fetch_pandas_all()

        print(f"Rows fetched: {len(df)}")
        print("Query execution completed")
        return df

    finally:
        cursor.close()
```

```python
from snowflake.connector.pandas_tools import write_pandas
import json

def load_to_target(df, table_name):
    print("Preparing data for load...")

    # Clean JSON columns
    for col in df.columns:
        if df[col].dtype == "object":
            # Remove unwanted newline characters
            df[col] = df[col].apply(
                lambda x: x.replace("\n", "") if isinstance(x, str) else x
            )

            # Try converting valid JSON strings to dict
            def try_json(val):
                if isinstance(val, str):
                    try:
                        return json.loads(val)
                    except:
                        return val
                return val

            df[col] = df[col].apply(try_json)

    print("Loading to target...")

    success, nchunks, nrows, _ = write_pandas(
        conn=sink_conn,
        df=df,
        table_name=table_name
    )

    if success:
        print(f"Successfully loaded {nrows} rows into {table_name}")
    else:
        print("Load failed")

    return success
```
